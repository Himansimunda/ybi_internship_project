```python
import pandas as pd
```

```python
data = {
    "Name": ["Sita", "Gita", "Raju", "Rani", "Amit", "Rahul", "Priya", "Kiran"],
    "Study_Hours": [2, 4, 1, 5, 3, 6, 4, 2],
    "Attendance": [70, 85, 60, 95, 80, 90, 88, 72],
    "Previous_Marks": [65, 75, 50, 90, 70, 92, 84, 68]
}

df = pd.DataFrame(data)

df
```

```python
average_study_hours = df["Study_Hours"].mean()

print("Average Study Hours:", average_study_hours)
```

```python
average_attendance = df["Attendance"].mean()

print("Average Attendance:", average_attendance)
```

```python
average_marks = df["Previous_Marks"].mean()

print("Average Previous Marks:", average_marks)
```

```python
highest_marks = df["Previous_Marks"].max()

print("Highest Previous Marks:", highest_marks)
```

```python
lowest_marks = df["Previous_Marks"].min()

print("Lowest Previous Marks:", lowest_marks)
```
