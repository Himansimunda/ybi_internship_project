```python
# Day 34: Feature Engineering Validation and Final Dataset Preparation
# Project: Student Performance Prediction System

import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

print("=" * 65)
print("STUDENT PERFORMANCE PREDICTION SYSTEM")
print("Day 34: Final Dataset Preparation")
print("=" * 65)

# Current notebook directory
PROJECT_DIR = Path.cwd()

print("Notebook directory:", PROJECT_DIR)
print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)

print("\nSetup completed successfully.")
```

```python
# Cell 2: Load Day 33 Engineered Features

from pathlib import Path
import pandas as pd

file_name = "day33_engineered_student_features.csv"

# Search common project locations
possible_paths = [
    Path.cwd() / file_name,
    Path.cwd() / "Data" / file_name,
    Path(r"C:\Internship_project") / file_name,
    Path(r"C:\Internship_project\Data") / file_name,
    Path(r"C:\Internship_project\Data\Data") / file_name,
]

file_path = next(
    (path for path in possible_paths if path.exists()),
    None
)

if file_path is None:
    print(f"ERROR: {file_name} was not found.")
    print("Checked locations:")
    for path in possible_paths:
        print(" -", path)
else:
    df = pd.read_csv(file_path)

    print("Dataset loaded successfully!")
    print("File location:", file_path)
    print("Dataset shape:", df.shape)

    display(df.head())
```

```python
# Cell 3: Dataset Structure and Quality Check
# Project: Student Performance Prediction System

if "df" not in globals():
    print("Dataset load hoinahi. Prathame Cell 2 run kara.")
else:
    print("=" * 60)
    print("DATASET STRUCTURE")
    print("=" * 60)

    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])

    print("\nColumn Names:")
    print(df.columns.tolist())

    print("\nData Types:")
    display(df.dtypes.to_frame(name="Data Type"))

    print("\nMissing Values:")
    missing_summary = df.isnull().sum().to_frame(name="Missing Count")
    missing_summary["Missing Percentage"] = (
        df.isnull().mean() * 100
    ).round(2)

    display(missing_summary)

    print("\nDuplicate Rows:", df.duplicated().sum())

    print("\nNumerical Summary:")
    display(df.describe().T)
```

```python
# Cell 4: Validate Required Features
# Project: Student Performance Prediction System

if "df" not in globals():
    print("Dataset load hoinahi. Prathame Cell 2 run kara.")

else:
    required_features = [
        "study_hours",
        "attendance",
        "previous_score",
        "assignments_completed"
    ]

    print("=" * 60)
    print("REQUIRED FEATURE VALIDATION")
    print("=" * 60)

    available_features = [
        col for col in required_features if col in df.columns
    ]

    missing_features = [
        col for col in required_features if col not in df.columns
    ]

    print("\nAvailable required features:")
    for feature in available_features:
        print(f"  [OK] {feature}")

    print("\nMissing required features:")
    if missing_features:
        for feature in missing_features:
            print(f"  [MISSING] {feature}")
    else:
        print("  None")

    if not missing_features:
        print("\nAll required input features are present.")
    else:
        print(
            "\nValidation needs attention. Check the actual column "
            "names before preparing the final dataset."
        )

    print("\nOther dataset columns:")
    other_columns = [
        col for col in df.columns
        if col not in required_features
    ]
    print(other_columns)
```

```python
# Cell 5: Numeric Feature Validation
# Project: Student Performance Prediction System

if "df" not in globals():
    print("Dataset load hoinahi. Prathame Cell 2 run kara.")

elif not all(
    col in df.columns
    for col in [
        "study_hours",
        "attendance",
        "previous_score",
        "assignments_completed"
    ]
):
    print("Required features missing achhi. Prathame Cell 4 ra output check kara.")

else:
    required_features = [
        "study_hours",
        "attendance",
        "previous_score",
        "assignments_completed"
    ]

    # Copy only the required input features
    X_final = df[required_features].copy()

    # Convert feature values to numeric; invalid values become NaN
    for col in required_features:
        X_final[col] = pd.to_numeric(
            X_final[col],
            errors="coerce"
        )

    # Replace positive/negative infinity with NaN
    X_final = X_final.replace([np.inf, -np.inf], np.nan)

    print("Numeric conversion completed.")
    print("\nFeature data types:")
    display(X_final.dtypes.to_frame(name="Data Type"))

    print("\nMissing or invalid values after conversion:")
    display(X_final.isna().sum().to_frame(name="Missing/Invalid Count"))

    print("\nPreview of input features:")
    display(X_final.head())

    print("\nPrepared feature dataset shape:", X_final.shape)
```

```python
# Cell 6: Feature Range Validation
# Project: Student Performance Prediction System

if "X_final" not in globals():
    print("X_final miluni. Prathame Cell 5 run kara.")

else:
    print("=" * 65)
    print("FEATURE RANGE VALIDATION")
    print("=" * 65)

    # Assumptions based on common student-data formats.
    # Confirm these ranges against your dataset's documentation.
    range_rules = {
        "study_hours": (0, 24),
        "attendance": (0, 100),
        "previous_score": (0, 100),
        "assignments_completed": (0, None)
    }

    validation_results = []

    for feature, (minimum, maximum) in range_rules.items():
        if feature not in X_final.columns:
            continue

        values = X_final[feature]
        valid_values = values.dropna()

        below_min = (
            int((valid_values < minimum).sum())
            if minimum is not None else 0
        )

        above_max = (
            int((valid_values > maximum).sum())
            if maximum is not None else 0
        )

        non_integer = 0
        if feature == "assignments_completed":
            non_integer = int(
                ((valid_values % 1) != 0).sum()
            )

        validation_results.append({
            "Feature": feature,
            "Minimum Allowed": minimum,
            "Maximum Allowed": (
                maximum if maximum is not None else "No limit specified"
            ),
            "Missing Values": int(values.isna().sum()),
            "Below Minimum": below_min,
            "Above Maximum": above_max,
            "Non-Integer Values": non_integer
        })

    range_report = pd.DataFrame(validation_results)

    display(range_report)

    print("\nNote:")
    print("- Range limits are assumptions and should be verified.")
    print("- Missing values are not automatically removed.")
    print("- Investigate unusual values before changing the dataset.")
```

```python
# Cell 7: Missing Value Analysis
# Project: Student Performance Prediction System

if "X_final" not in globals():
    print("X_final miluni. Prathame Cell 5 run kara.")

else:
    print("=" * 65)
    print("MISSING VALUE ANALYSIS")
    print("=" * 65)

    missing_report = pd.DataFrame({
        "Feature": X_final.columns,
        "Total Rows": len(X_final),
        "Missing Count": X_final.isna().sum().values,
        "Missing Percentage": (
            X_final.isna().mean().values * 100
        ).round(2),
        "Available Count": X_final.notna().sum().values
    })

    display(missing_report)

    total_missing = int(X_final.isna().sum().sum())
    total_cells = X_final.shape[0] * X_final.shape[1]

    print("\nTotal missing/invalid cells:", total_missing)
    print("Total feature cells:", total_cells)

    if total_cells > 0:
        print(
            "Overall missing percentage:",
            round(total_missing / total_cells * 100, 2),
            "%"
        )

    if total_missing == 0:
        print("\nAll selected features have complete values.")
    else:
        print(
            "\nMissing values found. Investigate their source before "
            "choosing an imputation or removal strategy."
        )

    print("\nDataset shape remains unchanged:", X_final.shape)
```

```python
# Cell 8: Duplicate Record Analysis
# Project: Student Performance Prediction System

if "X_final" not in globals():
    print("X_final miluni. Prathame Cell 5 run kara.")

else:
    print("=" * 65)
    print("DUPLICATE RECORD ANALYSIS")
    print("=" * 65)

    duplicate_mask = X_final.duplicated(keep=False)
    duplicate_records = X_final.loc[duplicate_mask].copy()

    duplicate_count = int(X_final.duplicated().sum())
    affected_rows = int(duplicate_mask.sum())

    print("Total rows:", len(X_final))
    print("Duplicate rows after first occurrence:", duplicate_count)
    print("Total rows belonging to duplicate groups:", affected_rows)

    if duplicate_records.empty:
        print("\nNo duplicate feature records found.")
    else:
        print("\nSample duplicate feature records:")
        display(duplicate_records.head(10))

    duplicate_summary = pd.DataFrame({
        "Metric": [
            "Total rows",
            "Duplicate rows after first occurrence",
            "Rows belonging to duplicate groups"
        ],
        "Count": [
            len(X_final),
            duplicate_count,
            affected_rows
        ]
    })

    print("\nDuplicate Summary:")
    display(duplicate_summary)

    print("\nOriginal prepared feature dataset remains unchanged.")
```

```python
# Cell 9: Final Dataset Quality Report
# Project: Student Performance Prediction System

if "X_final" not in globals():
    print("X_final miluni. Prathame Cell 5 run kara.")

else:
    print("=" * 70)
    print("FINAL DATASET QUALITY REPORT")
    print("=" * 70)

    quality_rows = []

    for col in X_final.columns:
        series = X_final[col]
        available = series.dropna()

        quality_rows.append({
            "Feature": col,
            "Data Type": str(series.dtype),
            "Total Rows": len(series),
            "Missing Values": int(series.isna().sum()),
            "Unique Values": int(series.nunique(dropna=True)),
            "Minimum": (
                available.min() if not available.empty else np.nan
            ),
            "Maximum": (
                available.max() if not available.empty else np.nan
            ),
            "Mean": (
                round(available.mean(), 2)
                if not available.empty else np.nan
            )
        })

    quality_report = pd.DataFrame(quality_rows)

    display(quality_report)

    print("\nOverall Summary")
    print("-" * 40)
    print("Number of rows:", X_final.shape[0])
    print("Number of input features:", X_final.shape[1])
    print("Total missing cells:", int(X_final.isna().sum().sum()))
    print("Exact duplicate feature rows:", int(X_final.duplicated().sum()))

    if X_final.isna().sum().sum() == 0:
        print("\nStatus: No missing values in selected input features.")
    else:
        print("\nStatus: Missing values need further investigation.")

    print("\nQuality report generated successfully.")
```

```python
# Cell 10: Prepare Final Feature Dataset
# Project: Student Performance Prediction System

if "X_final" not in globals():
    print("X_final miluni. Prathame Cell 5 run kara.")

else:
    expected_features = [
        "study_hours",
        "attendance",
        "previous_score",
        "assignments_completed"
    ]

    missing_columns = [
        col for col in expected_features
        if col not in X_final.columns
    ]

    if missing_columns:
        print("Cannot prepare dataset. Missing columns:", missing_columns)

    else:
        # Select features in a consistent order
        final_features = X_final[expected_features].copy()

        # Check missing values before saving
        missing_count = int(final_features.isna().sum().sum())

        print("=" * 65)
        print("FINAL FEATURE DATASET PREPARATION")
        print("=" * 65)

        print("Dataset shape:", final_features.shape)
        print("Input features:", final_features.columns.tolist())
        print("Missing values:", missing_count)

        if missing_count > 0:
            print(
                "\nWARNING: Missing or invalid values exist. "
                "Dataset is prepared for inspection, but it is not "
                "ready for model training until these are handled."
            )
        else:
            print("\nAll selected input features have values.")

        print("\nFinal feature preview:")
        display(final_features.head())

        print("\nFinal feature data types:")
        display(final_features.dtypes.to_frame(name="Data Type"))
```

```python
# Cell 11: Save Final Dataset and Quality Reports
# Project: Student Performance Prediction System

if "final_features" not in globals():
    print("Final dataset miluni. Prathame Cell 10 run kara.")

elif "quality_report" not in globals():
    print("Quality report miluni. Prathame Cell 9 run kara.")

else:
    output_dir = Path.cwd()
    output_dir.mkdir(parents=True, exist_ok=True)

    # Output file paths
    dataset_path = output_dir / "day34_final_student_features.csv"
    quality_path = output_dir / "day34_dataset_quality_report.csv"
    missing_path = output_dir / "day34_missing_value_report.csv"

    # Prepare missing-value report
    missing_report = pd.DataFrame({
        "Feature": final_features.columns,
        "Missing Count": final_features.isna().sum().values,
        "Missing Percentage": (
            final_features.isna().mean().values * 100
        ).round(2)
    })

    # Save outputs
    final_features.to_csv(dataset_path, index=False)
    quality_report.to_csv(quality_path, index=False)
    missing_report.to_csv(missing_path, index=False)

    print("=" * 65)
    print("DAY 34 OUTPUT FILES")
    print("=" * 65)

    for path in [dataset_path, quality_path, missing_path]:
        print(f"Saved: {path}")
        print(f"File size: {path.stat().st_size:,} bytes\n")

    print("Final dataset shape:", final_features.shape)
    print("Total missing values:", int(final_features.isna().sum().sum()))

    if final_features.isna().sum().sum() > 0:
        print(
            "WARNING: Dataset saved with missing values. "
            "Handle them before model training."
        )
    else:
        print("All selected input features are complete.")

    print("\nOutput files saved successfully.")
```
