```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Name": ["Sita", "Gita", "Raju", "Rani", "Amit", "Rahul", "Priya", "Kiran"],
    "Study_Hours": [2, 4, 1, 5, 3, 6, 4, 2],
    "Attendance": [70, 85, 60, 95, 80, 90, 88, 72],
    "Previous_Marks": [65, 75, 50, 90, 70, 92, 84, 68]
}

df = pd.DataFrame(data)

df
```

```python
print("Average Study Hours:", df["Study_Hours"].mean())
print("Maximum Study Hours:", df["Study_Hours"].max())
print("Minimum Study Hours:", df["Study_Hours"].min())
```

```python
print("Average Attendance:", df["Attendance"].mean())
print("Highest Attendance:", df["Attendance"].max())
print("Lowest Attendance:", df["Attendance"].min())
```

```python
print("Average Previous Marks:", df["Previous_Marks"].mean())
print("Highest Previous Marks:", df["Previous_Marks"].max())
print("Lowest Previous Marks:", df["Previous_Marks"].min())
```

```python
plt.figure(figsize=(8, 5))

plt.bar(df["Name"], df["Study_Hours"])

plt.xlabel("Student")
plt.ylabel("Study Hours")
plt.title("Study Hours of Students")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.bar(df["Name"], df["Attendance"])

plt.xlabel("Student")
plt.ylabel("Attendance (%)")
plt.title("Attendance of Students")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Study_Hours"], df["Previous_Marks"])

plt.xlabel("Study Hours")
plt.ylabel("Previous Marks")
plt.title("Study Hours vs Previous Marks")

plt.show()
```

```python
correlation = df["Study_Hours"].corr(df["Previous_Marks"])

print("Correlation between Study Hours and Previous Marks:", correlation)
```
