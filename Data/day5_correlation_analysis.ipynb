```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Name": ["Sita", "Gita", "Raju", "Rani", "Amit", "Rahul", "Priya", "Kiran"],
    "Study_Hours": [2, 4, 1, 5, 3, 6, 4, 2],
    "Attendance": [70, 85, 60, 95, 80, 90, 88, 72],
    "Previous_Marks": [65, 75, 50, 90, 70, 92, 84, 68]
}

df = pd.DataFrame(data)

df
```

```python
numeric_data = df[["Study_Hours", "Attendance", "Previous_Marks"]]

numeric_data
```

```python
correlation_matrix = numeric_data.corr()

correlation_matrix
```

```python
study_marks_correlation = df["Study_Hours"].corr(df["Previous_Marks"])

print("Study Hours vs Previous Marks:", study_marks_correlation)
```

```python
attendance_marks_correlation = df["Attendance"].corr(df["Previous_Marks"])

print("Attendance vs Previous Marks:", attendance_marks_correlation)
```

```python
study_attendance_correlation = df["Study_Hours"].corr(df["Attendance"])

print("Study Hours vs Attendance:", study_attendance_correlation)
```

```python
plt.figure(figsize=(8, 5))

plt.imshow(correlation_matrix)

plt.xticks(range(len(correlation_matrix.columns)), correlation_matrix.columns)
plt.yticks(range(len(correlation_matrix.columns)), correlation_matrix.columns)

plt.colorbar()

plt.title("Correlation Matrix")

plt.show()
```

```python
print("Correlation Matrix:")
print(correlation_matrix)

print("\nStudy Hours vs Previous Marks:")
print(study_marks_correlation)

print("\nAttendance vs Previous Marks:")
print(attendance_marks_correlation)
```
