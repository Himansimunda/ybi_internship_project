```python
import pandas as pd
from pathlib import Path

print("Student Performance Prediction System")
print("Day 36: Model Readiness Check")
```

```python
root = Path(r"C:\Internship_project")
files = list(root.rglob("day35_validated_student_features.csv"))

if files:
    df = pd.read_csv(files[0])
    print("Dataset loaded:", files[0])
    print("Dataset shape:", df.shape)
    display(df.head())
else:
    print("Dataset not found. Day 35 CSV file check kara.")
```

```python
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
required = [
    "study_hours",
    "attendance",
    "previous_score",
    "assignments_completed"
]

missing = [col for col in required if col not in df.columns]

if not missing:
    print("All required features are available!")
else:
    print("Missing features:", missing)
```

```python
if all(col in df.columns for col in required):
    display(df[required].describe().T)
else:
    print("Required features check.")
```

```python
report = pd.DataFrame({
    "Check": [
        "Dataset Loaded",
        "Required Features Available",
        "Missing Values",
        "Duplicate Rows"
    ],
    "Result": [
        not df.empty,
        all(col in df.columns for col in required),
        int(df.isnull().sum().sum()),
        int(df.duplicated().sum())
    ]
})

report.to_csv("day36_model_readiness_report.csv", index=False)
display(report)

print("Day 36 report saved successfully!")
```

```python
output = "day36_model_ready_dataset.csv"

df.to_csv(output, index=False)

print("Dataset saved:", output)
print("Final shape:", df.shape)
print("Day 36 completed!")
```
