```python
import pandas as pd
from pathlib import Path
from datetime import date
```

```python
project_name = "Student Performance Prediction System"

summary = {
    "Project Name": project_name,
    "Duration": "45 Days",
    "Final Day": "Day 45",
    "Main Activities": "Data cleaning, preprocessing, feature engineering, "
                       "dataset validation and model readiness",
    "Model Status": "Pending actual performance labels",
    "Completion Date": str(date.today())
}

display(pd.DataFrame(summary.items(), columns=["Item", "Details"]))
```

```python
project_dir = Path(r"C:\Internship_project")

output_files = [
    "day33_engineered_student_features.csv",
    "day35_validated_student_features.csv",
    "day36_model_ready_dataset.csv",
    "day40_prepared_student_features.csv",
    "day41_validated_training_features.csv",
    "day44_model_evaluation_readiness_report.csv"
]

file_status = []

for filename in output_files:
    found = list(project_dir.rglob(filename))
    file_status.append({
        "File": filename,
        "Status": "Available" if found else "Not Found"
    })

file_report = pd.DataFrame(file_status)
display(file_report)
```

```python
final_status = pd.DataFrame({
    "Task": [
        "Data Preparation",
        "Data Validation",
        "Feature Engineering",
        "Model Readiness",
        "Model Evaluation"
    ],
    "Status": [
        "Completed",
        "Completed",
        "Completed",
        "Completed",
        "Pending actual performance labels"
    ]
})

display(final_status)
```

```python
report_path = Path.cwd()

file_report.to_csv(
    report_path / "day45_project_file_report.csv",
    index=False
)

final_status.to_csv(
    report_path / "day45_final_project_status.csv",
    index=False
)

print("Final project reports saved successfully!")
```
