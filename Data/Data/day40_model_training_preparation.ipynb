```python
import pandas as pd
from pathlib import Path

print("Student Performance Prediction System")
print("Day 40: Model Training Preparation")
```

```python
root = Path(r"C:\Internship_project")
files = list(root.rglob("day39_analyzed_student_features.csv"))

if files:
    df = pd.read_csv(files[0])
    print("Dataset loaded:", files[0])
    print("Shape:", df.shape)
    display(df.head())
else:
    print("Dataset not found. Check Day 39 CSV file.")
```

```python
features = [
    "study_hours",
    "attendance",
    "previous_score",
    "assignments_completed"
]

available_features = [
    col for col in features if col in df.columns
]

X = df[available_features]

print("Input features:", available_features)
print("Feature shape:", X.shape)
display(X.head())
```

```python
target = "predicted_performance"

if target in df.columns:
    print("Target column found:", target)
    print(df[target].value_counts(dropna=False))
else:
    print("Target column not found.")
```

```python
if target in df.columns:
    print("Target data type:", df[target].dtype)
    print("Unique target values:", df[target].nunique())
    print("Missing target values:", df[target].isnull().sum())

    print("\nImportant:")
    print("Confirm whether this column contains actual outcomes")
    print("or predictions generated by an earlier model.")
```

```python
feature_report = pd.DataFrame({
    "Feature": available_features,
    "Data Type": [df[col].dtype for col in available_features],
    "Missing Values": [
        df[col].isnull().sum() for col in available_features
    ]
})

feature_report.to_csv(
    "day40_feature_preparation_report.csv",
    index=False
)

display(feature_report)
print("Feature report saved!")
```

```python
X.to_csv("day40_prepared_student_features.csv", index=False)

print("Prepared features saved successfully!")
print("Records:", len(X))
print("Features:", len(X.columns))
```
