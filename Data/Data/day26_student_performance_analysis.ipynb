```python
# Day 26 - Student Performance Prediction System

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

print("=" * 50)
print("STUDENT PERFORMANCE PREDICTION SYSTEM")
print("DAY 26: STUDENT PERFORMANCE ANALYSIS")
print("=" * 50)

print("\nLibraries imported successfully!")
print("Day 26 setup completed.")
```

```python
from pathlib import Path
import pandas as pd

# Check the parent folder
file_path = Path.cwd().parent / "day23_student_performance_predictions.csv"

if file_path.exists():
    df = pd.read_csv(file_path)

    print("Dataset loaded successfully!")
    print("Dataset shape:", df.shape)
    display(df.head())
else:
    print("CSV file not found!")
    print("Current folder:", Path.cwd())
    print("Parent folder:", Path.cwd().parent)
    print("CSV files in parent folder:")
    print(list(Path.cwd().parent.glob("*.csv")))
```

```python
# Cell 3: Explore Dataset Information

print("=" * 50)
print("DATASET INFORMATION")
print("=" * 50)

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nStatistical Summary:")
display(df.describe(include="all"))

print("\nDataset exploration completed!")
```

```python
# Cell 4: Student Performance Analysis

print("=" * 50)
print("STUDENT PERFORMANCE ANALYSIS")
print("=" * 50)

print("\nAverage Study Hours:")
if "study_hours" in df.columns:
    print(round(df["study_hours"].mean(), 2))

print("\nAverage Attendance:")
if "attendance" in df.columns:
    print(round(df["attendance"].mean(), 2))

print("\nAverage Previous Score:")
if "previous_score" in df.columns:
    print(round(df["previous_score"].mean(), 2))

print("\nAverage Assignments Completed:")
if "assignments_completed" in df.columns:
    print(round(df["assignments_completed"].mean(), 2))

print("\nPredicted Performance Distribution:")
if "predicted_performance" in df.columns:
    print(df["predicted_performance"].value_counts())

print("\nStudent performance analysis completed!")
```

```python
# Cell 5: Study Hours Visualization

import matplotlib.pyplot as plt

if "study_hours" in df.columns:
    plt.figure(figsize=(8, 5))

    plt.hist(df["study_hours"].dropna(), bins=10, edgecolor="black")

    plt.title("Distribution of Student Study Hours")
    plt.xlabel("Study Hours")
    plt.ylabel("Number of Students")

    plt.grid(axis="y", alpha=0.3)
    plt.tight_layout()
    plt.show()

    print("Study hours visualization completed!")

else:
    print("study_hours column not found.")
```

```python
# Cell 6: Attendance Analysis

import matplotlib.pyplot as plt

if "attendance" in df.columns:
    plt.figure(figsize=(8, 5))

    plt.hist(
        df["attendance"].dropna(),
        bins=10,
        edgecolor="black"
    )

    plt.title("Distribution of Student Attendance")
    plt.xlabel("Attendance")
    plt.ylabel("Number of Students")

    plt.grid(axis="y", alpha=0.3)
    plt.tight_layout()
    plt.show()

    print("Attendance analysis completed!")

else:
    print("attendance column not found.")
```

```python
# Cell 7: Previous Score Analysis

import matplotlib.pyplot as plt

if "previous_score" in df.columns:
    scores = df["previous_score"].dropna()

    print("Previous Score Analysis")
    print("-" * 35)
    print("Average Score:", round(scores.mean(), 2))
    print("Highest Score:", scores.max())
    print("Lowest Score:", scores.min())

    plt.figure(figsize=(8, 5))
    plt.hist(scores, bins=10, edgecolor="black")

    plt.title("Distribution of Previous Student Scores")
    plt.xlabel("Previous Score")
    plt.ylabel("Number of Students")

    plt.grid(axis="y", alpha=0.3)
    plt.tight_layout()
    plt.show()

else:
    print("previous_score column not found.")
```

```python
# Cell 8: Correlation Analysis

import matplotlib.pyplot as plt
import numpy as np

numeric_df = df.select_dtypes(include="number")
correlation_matrix = numeric_df.corr()

print("Feature Correlation Matrix:")
display(correlation_matrix.round(2))

plt.figure(figsize=(9, 6))

plt.imshow(correlation_matrix, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar(label="Correlation")

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

plt.title("Student Performance Feature Correlation")
plt.tight_layout()
plt.show()

print("Correlation analysis completed!")
```

```python
# Cell 9: Assignment Completion Analysis

import matplotlib.pyplot as plt

if "assignments_completed" in df.columns:

    assignments = df["assignments_completed"].dropna()

    print("ASSIGNMENT COMPLETION ANALYSIS")
    print("-" * 40)
    print("Average Assignments Completed:", round(assignments.mean(), 2))
    print("Maximum Assignments Completed:", assignments.max())
    print("Minimum Assignments Completed:", assignments.min())

    plt.figure(figsize=(8, 5))
    plt.hist(assignments, bins=10, edgecolor="black")

    plt.title("Distribution of Assignments Completed")
    plt.xlabel("Assignments Completed")
    plt.ylabel("Number of Students")

    plt.grid(axis="y", alpha=0.3)
    plt.tight_layout()
    plt.show()

    print("Assignment analysis completed!")

else:
    print("assignments_completed column not found.")
```

```python
# Cell 10: Save Analysis Results

from pathlib import Path

output_dir = Path.cwd()

# Save dataset summary
summary = df.describe(include="all").transpose()
summary.to_csv(output_dir / "day26_student_performance_summary.csv")

# Save data quality report
quality_report = pd.DataFrame({
    "Missing_Values": df.isnull().sum(),
    "Data_Type": df.dtypes.astype(str),
    "Unique_Values": df.nunique()
})

quality_report.to_csv(
    output_dir / "day26_dataset_quality_report.csv"
)

# Save correlation matrix
numeric_df = df.select_dtypes(include="number")

if not numeric_df.empty:
    numeric_df.corr().to_csv(
        output_dir / "day26_correlation_matrix.csv"
    )

print("Analysis results saved successfully!")
print("Output folder:", output_dir)
print("\nSaved files:")

for file in output_dir.glob("day26_*.csv"):
    print("-", file.name)
```
