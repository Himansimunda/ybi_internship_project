```python
import pandas as pd
from pathlib import Path

print("Student Performance Prediction System")
print("Day 41: Training Dataset Validation")
```

```python
root = Path(r"C:\Internship_project")
files = list(root.rglob("day40_prepared_student_features.csv"))

if files:
    X = pd.read_csv(files[0])
    print("Dataset loaded:", files[0])
    print("Shape:", X.shape)
    display(X.head())
else:
    print("Day 40 dataset not found. Check CSV file.")
```

```python
required = [
    "study_hours",
    "attendance",
    "previous_score",
    "assignments_completed"
]

missing = [col for col in required if col not in X.columns]

print("Missing columns:", missing)
print("Available columns:", list(X.columns))
```

```python
print("Missing values:\n", X.isnull().sum())
print("\nDuplicate rows:", X.duplicated().sum())
```

```python
numeric_check = X[required].select_dtypes(include="number")

print("Numeric feature count:", numeric_check.shape[1])
display(numeric_check.describe().T)
```

```python
validation_report = pd.DataFrame({
    "Feature": required,
    "Data Type": [X[col].dtype for col in required],
    "Missing Values": [X[col].isnull().sum() for col in required],
    "Unique Values": [X[col].nunique() for col in required]
})

display(validation_report)

validation_report.to_csv(
    "day41_training_validation_report.csv",
    index=False
)

print("Validation report saved!")
```

```python
X.to_csv("day41_validated_training_features.csv", index=False)

print("Validated dataset saved!")
print("Records:", len(X))
print("Features:", len(X.columns))
```
