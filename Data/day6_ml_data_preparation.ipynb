```python
import pandas as pd
```

```python
data = {
    "Name": ["Sita", "Gita", "Raju", "Rani", "Amit", "Rahul", "Priya", "Kiran"],
    "Study_Hours": [2, 4, 1, 5, 3, 6, 4, 2],
    "Attendance": [70, 85, 60, 95, 80, 90, 88, 72],
    "Previous_Marks": [65, 75, 50, 90, 70, 92, 84, 68]
}

df = pd.DataFrame(data)

df
```

```python
import sklearn
print("Scikit-learn is working!")
```

```python
from sklearn.model_selection import train_test_split

X = df[["Study_Hours", "Attendance", "Previous_Marks"]]

df["Final_Marks"] = [68, 78, 55, 94, 74, 95, 88, 71]

y = df["Final_Marks"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

print("Data split successfully!")
```

```python
print("X_train:")
print(X_train)

print("\ny_train:")
print(y_train)
```

```python
print("X_test:")
print(X_test)

print("\ny_test:")
print(y_test)
```
