```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
data = {
    "Study_Hours": [1, 2, 3, 4, 5, 6, 7, 8],
    "Attendance": [60, 65, 70, 75, 80, 85, 90, 95],
    "Previous_Marks": [45, 50, 55, 60, 68, 75, 82, 90],
    "Result": ["Fail", "Fail", "Fail", "Pass", "Pass", "Pass", "Pass", "Pass"]
}

df = pd.DataFrame(data)

df
```

```python
df["Result"] = df["Result"].map({
    "Fail": 0,
    "Pass": 1
})

df
```

```python
X = df[["Study_Hours", "Attendance", "Previous_Marks"]]
y = df["Result"]

print("Features:")
print(X)

print("\nTarget:")
print(y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

print("Training data:")
print(X_train)

print("\nTesting data:")
print(X_test)

print("\nTraining target:")
print(y_train)

print("\nTesting target:")
print(y_test)
```

```python
data = {
    "Study_Hours": [1, 2, 3, 4, 5, 6, 7, 8],
    "Attendance": [60, 65, 70, 75, 80, 85, 90, 95],
    "Previous_Marks": [45, 50, 55, 60, 68, 75, 82, 90],
    "Result": [0, 0, 0, 1, 1, 1, 1, 1]
}

df = pd.DataFrame(data)

print(df)
print("\nShape:", df.shape)
```

```python
X = df[["Study_Hours", "Attendance", "Previous_Marks"]]
y = df["Result"]

print("X shape:", X.shape)
print("y shape:", y.shape)
print("\ny values:")
print(y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)

print("\nTraining target values:")
print(y_train)
```

```python
model = LogisticRegression()

model.fit(X_train, y_train)

print("Model training completed!")
```

```python
y_pred = model.predict(X_test)

print("Predicted Results:")
print(y_pred)
```

```python
comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

print(comparison)
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", accuracy)
print("Accuracy Percentage:", accuracy * 100, "%")
```

```python
new_student = pd.DataFrame({
    "Study_Hours": [5],
    "Attendance": [82],
    "Previous_Marks": [70]
})

prediction = model.predict(new_student)

if prediction[0] == 1:
    print("Predicted Result: Pass")
else:
    print("Predicted Result: Fail")
```

```python
probability = model.predict_proba(new_student)

print("Fail Probability:", probability[0][0])
print("Pass Probability:", probability[0][1])
```
