```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

print("Day 9 notebook working!")
```

```python
data = {
    "Study_Hours": [1, 2, 3, 4, 5, 6, 7, 8],
    "Final_Marks": [50, 55, 62, 70, 80, 88, 91, 95]
}

df = pd.DataFrame(data)

df
```

```python
X = df[["Study_Hours"]]
y = df["Final_Marks"]

print("Feature:")
print(X)

print("\nTarget:")
print(y)
```

```python
poly = PolynomialFeatures(degree=2)

X_poly = poly.fit_transform(X)

print("Original X:")
print(X)

print("\nPolynomial X:")
print(X_poly)
```

```python
model = LinearRegression()

model.fit(X_poly, y)

print("Polynomial Regression model trained successfully!")
```

```python
y_pred = model.predict(X_poly)

print("Predicted Final Marks:")
print(y_pred)
```

```python
plt.scatter(X, y, label="Actual Data")

plt.plot(X, y_pred, label="Polynomial Regression")

plt.xlabel("Study Hours")
plt.ylabel("Final Marks")
plt.title("Polynomial Regression")
plt.legend()

plt.show()
```

```python
new_student = pd.DataFrame({
    "Study_Hours": [6.5]
})

new_student_poly = poly.transform(new_student)

predicted_marks = model.predict(new_student_poly)

print("Study Hours:", new_student["Study_Hours"].iloc[0])
print("Predicted Final Marks:", predicted_marks[0])
```

```python
print("Model Coefficients:", model.coef_)
print("Model Intercept:", model.intercept_)
```

```python
print("Day 9 Polynomial Regression completed!")
print("Degree used:", poly.degree)
print("Feature used: Study_Hours")
print("Target: Final_Marks")
print("New student prediction completed.")
```
