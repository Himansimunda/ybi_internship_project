```python
# Day 23: Student Performance Prediction System

import pandas as pd
import joblib

print("Libraries imported successfully!")
```

```python
# Load the trained Random Forest model

model_path = r"C:\Internship_project\models\day22_student_performance_model.pkl"

model = joblib.load(model_path)

print("Student Performance Prediction Model loaded successfully!")
```

```python
# Create input data for a new student

new_student = pd.DataFrame({
    "study_hours": [6],
    "attendance": [90],
    "previous_score": [78],
    "assignments_completed": [9]
})

display(new_student)
```

```python
# Predict student performance

prediction = model.predict(new_student)

print("Student Performance Prediction System")
print("-------------------------------------")
print("Predicted Performance:", prediction[0])
```

```python
# Save student data and prediction to a CSV file

new_student["predicted_performance"] = prediction

output_path = r"C:\Internship_project\Data\day23_student_performance_predictions.csv"

new_student.to_csv(output_path, index=False)

print("Prediction saved successfully!")
print("File location:", output_path)
```

```python
# Display the saved prediction results

results_path = r"C:\Internship_project\Data\day23_student_performance_predictions.csv"

results = pd.read_csv(results_path)

print("Student Performance Prediction Results")
print("--------------------------------------")

display(results)
```

```python
# Generate a student performance report

student = results.iloc[0]

report_path = r"C:\Internship_project\Data\day23_student_performance_report.txt"

report = f"""
STUDENT PERFORMANCE PREDICTION SYSTEM
=====================================

Student Details:
Study Hours: {student['study_hours']}
Attendance: {student['attendance']}%
Previous Score: {student['previous_score']}
Assignments Completed: {student['assignments_completed']}

Predicted Performance: {student['predicted_performance']}

Note:
This prediction is generated by a Random Forest model
trained on a small synthetic practice dataset.
It should not be treated as a definitive assessment
of a student's actual academic ability.
"""

with open(report_path, "w", encoding="utf-8") as file:
    file.write(report)

print("Student performance report saved successfully!")
print("Report location:", report_path)
```

```python
# Check the model's score on the new student input

print("Student Performance Prediction System")
print("-------------------------------------")
print("Prediction:", prediction[0])
print("Input data:")
display(new_student)
```

```python
# Visualize the predicted performance

import matplotlib.pyplot as plt

predicted_class = str(prediction[0])

plt.figure(figsize=(7, 4))
plt.bar(
    [predicted_class],
    [1]
)

plt.title("Student Performance Prediction")
plt.xlabel("Predicted Performance Class")
plt.ylabel("Prediction Count")
plt.ylim(0, 1.2)
plt.grid(axis="y", linestyle="--", alpha=0.3)
plt.show()
```
