```python
from pathlib import Path
import pandas as pd

current_dir = Path.cwd()

possible_paths = [
    current_dir / "day23_student_performance_predictions.csv",
    current_dir.parent / "day23_student_performance_predictions.csv",
    current_dir.parent.parent / "day23_student_performance_predictions.csv"
]

dataset_path = next(
    (path for path in possible_paths if path.exists()),
    None
)

if dataset_path is not None:
    df = pd.read_csv(dataset_path)
    print("Dataset loaded successfully!")
    print("File path:", dataset_path)
    print("Dataset shape:", df.shape)
    display(df.head())
else:
    print("Dataset file not found.")
    print("Current folder:", current_dir)
    print("Please check the CSV file location.")
```

```python
# Cell 3: Dataset Information

if "df" in globals() and not df.empty:
    print("Dataset Information")
    print("=" * 40)

    print("\n1. Dataset Shape:")
    print(df.shape)

    print("\n2. Column Names:")
    print(df.columns.tolist())

    print("\n3. Data Types:")
    print(df.dtypes)

    print("\n4. Missing Values:")
    print(df.isnull().sum())

    print("\n5. Duplicate Rows:")
    print(df.duplicated().sum())

    print("\n6. Statistical Summary:")
    display(df.describe(include="all"))
else:
    print("Dataset is not loaded. Please run Cell 2 first.")
```

```python
# Cell 4: Student Performance Statistics

if "df" in globals() and not df.empty:

    print("Student Performance Statistics")
    print("=" * 40)

    numeric_columns = df.select_dtypes(include="number").columns.tolist()

    if numeric_columns:
        print("\nAverage Values:")
        display(df[numeric_columns].mean().to_frame("Average"))

        print("\nMinimum Values:")
        display(df[numeric_columns].min().to_frame("Minimum"))

        print("\nMaximum Values:")
        display(df[numeric_columns].max().to_frame("Maximum"))

        print("\nMedian Values:")
        display(df[numeric_columns].median().to_frame("Median"))
    else:
        print("No numeric columns found.")

else:
    print("Dataset is not loaded. Please run Cell 2 first.")
```

```python
# Cell 5: Study Hours vs Predicted Performance

if "df" in globals() and not df.empty:

    required_columns = ["study_hours", "predicted_performance"]

    if all(col in df.columns for col in required_columns):

        plt.figure(figsize=(9, 5))

        plt.scatter(
            df["study_hours"],
            df["predicted_performance"],
            alpha=0.7
        )

        plt.title("Study Hours vs Predicted Performance")
        plt.xlabel("Study Hours")
        plt.ylabel("Predicted Performance")
        plt.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()

        print("Study hours vs predicted performance chart created successfully!")

    else:
        print("Required columns are missing:", required_columns)

else:
    print("Dataset is not loaded. Please run Cell 2 first.")
```

```python
# Cell 6: Attendance vs Predicted Performance

if "df" in globals() and not df.empty:

    required_columns = ["attendance", "predicted_performance"]

    if all(col in df.columns for col in required_columns):

        plt.figure(figsize=(9, 5))

        plt.scatter(
            df["attendance"],
            df["predicted_performance"],
            alpha=0.7
        )

        plt.title("Attendance vs Predicted Performance")
        plt.xlabel("Attendance")
        plt.ylabel("Predicted Performance")
        plt.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()

        print("Attendance vs predicted performance chart created successfully!")

    else:
        print("Required columns are missing:", required_columns)

else:
    print("Dataset is not loaded. Please run Cell 2 first.")
```

```python
# Cell 7: Previous Score vs Predicted Performance

if "df" in globals() and not df.empty:

    required_columns = ["previous_score", "predicted_performance"]

    if all(col in df.columns for col in required_columns):

        plt.figure(figsize=(9, 5))

        plt.scatter(
            df["previous_score"],
            df["predicted_performance"],
            alpha=0.7
        )

        plt.title("Previous Score vs Predicted Performance")
        plt.xlabel("Previous Score")
        plt.ylabel("Predicted Performance")
        plt.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()

        print("Previous score analysis completed successfully!")

    else:
        print("Required columns are missing:", required_columns)

else:
    print("Dataset is not loaded. Please run Cell 2 first.")
```

```python
# Cell 8: Correlation Analysis

if "df" in globals() and not df.empty:

    numeric_df = df.select_dtypes(include="number")

    if numeric_df.shape[1] >= 2:

        correlation_matrix = numeric_df.corr()

        print("Correlation Matrix:")
        display(correlation_matrix.round(2))

        plt.figure(figsize=(9, 6))

        plt.imshow(
            correlation_matrix,
            cmap="coolwarm",
            vmin=-1,
            vmax=1,
            aspect="auto"
        )

        plt.colorbar(label="Correlation Coefficient")

        plt.xticks(
            range(len(correlation_matrix.columns)),
            correlation_matrix.columns,
            rotation=45,
            ha="right"
        )

        plt.yticks(
            range(len(correlation_matrix.index)),
            correlation_matrix.index
        )

        plt.title("Student Performance Correlation Heatmap")

        for i in range(len(correlation_matrix.index)):
            for j in range(len(correlation_matrix.columns)):
                value = correlation_matrix.iloc[i, j]
                if pd.notna(value):
                    plt.text(
                        j, i, f"{value:.2f}",
                        ha="center",
                        va="center",
                        fontsize=9
                    )

        plt.tight_layout()
        plt.show()

        print("Correlation analysis completed successfully!")

    else:
        print("At least two numeric columns are required.")

else:
    print("Dataset is not loaded. Please run Cell 2 first.")
```

```python
# Cell 9: Assignments Completed Analysis

if "df" in globals() and not df.empty:

    if "assignments_completed" in df.columns:

        assignments = pd.to_numeric(
            df["assignments_completed"],
            errors="coerce"
        ).dropna()

        if not assignments.empty:

            print("Assignments Completed Analysis")
            print("=" * 40)

            print("Average assignments completed:", round(assignments.mean(), 2))
            print("Minimum assignments completed:", assignments.min())
            print("Maximum assignments completed:", assignments.max())

            plt.figure(figsize=(9, 5))

            plt.hist(
                assignments,
                bins="auto",
                edgecolor="black",
                alpha=0.8
            )

            plt.title("Distribution of Assignments Completed")
            plt.xlabel("Assignments Completed")
            plt.ylabel("Number of Students")
            plt.grid(axis="y", alpha=0.3)
            plt.tight_layout()
            plt.show()

            print("Assignments analysis completed successfully!")

        else:
            print("No valid numeric assignment data found.")

    else:
        print("'assignments_completed' column not found.")

else:
    print("Dataset is not loaded. Please run Cell 2 first.")
```

```python
# Cell 10: Generate Student Performance Summary

if "df" in globals() and not df.empty:

    numeric_df = df.select_dtypes(include="number")

    if not numeric_df.empty:

        summary = numeric_df.describe().T.round(2)

        print("Student Performance Summary")
        print("=" * 40)
        display(summary)

        output_path = Path.cwd() / "day27_student_performance_summary.csv"

        summary.to_csv(output_path, index=True)

        print("\nSummary saved successfully!")
        print("File path:", output_path.resolve())

    else:
        print("No numeric columns available for summary.")

else:
    print("Dataset is not loaded. Please run Cell 2 first.")
```

```python
# Cell 11: Dataset Quality Report

if "df" in globals() and not df.empty:

    quality_report = pd.DataFrame({
        "Column": df.columns,
        "Data_Type": [str(dtype) for dtype in df.dtypes],
        "Missing_Values": df.isnull().sum().values,
        "Missing_Percentage": (
            df.isnull().mean().values * 100
        ).round(2),
        "Unique_Values": df.nunique().values
    })

    print("Dataset Quality Report")
    print("=" * 40)
    display(quality_report)

    print("\nTotal rows:", len(df))
    print("Total columns:", len(df.columns))
    print("Duplicate rows:", df.duplicated().sum())

    output_path = Path.cwd() / "day27_dataset_quality_report.csv"

    quality_report.to_csv(output_path, index=False)

    print("\nQuality report saved successfully!")
    print("File path:", output_path.resolve())

else:
    print("Dataset is not loaded. Please run Cell 2 first.")
```

```python
# Cell 12: Save Correlation Matrix

if "df" in globals() and not df.empty:

    numeric_df = df.select_dtypes(include="number")

    if numeric_df.shape[1] >= 2:

        correlation_matrix = numeric_df.corr()

        print("Student Performance Correlation Matrix")
        display(correlation_matrix.round(2))

        output_path = Path.cwd() / "day27_correlation_matrix.csv"

        correlation_matrix.round(3).to_csv(output_path)

        print("\nCorrelation matrix saved successfully!")
        print("File path:", output_path.resolve())

    else:
        print("At least two numeric columns are required.")

else:
    print("Dataset is not loaded. Please run Cell 2 first.")
```

```python
# Cell 13: Generate Student Performance Analysis Report

if "df" in globals() and not df.empty:

    report_lines = []

    report_lines.append("STUDENT PERFORMANCE PREDICTION SYSTEM")
    report_lines.append("DAY 27: STUDENT PERFORMANCE ANALYSIS REPORT")
    report_lines.append("=" * 50)

    report_lines.append(f"\nTotal Records: {len(df)}")
    report_lines.append(f"Total Columns: {len(df.columns)}")
    report_lines.append(f"Duplicate Rows: {df.duplicated().sum()}")

    report_lines.append("\nMissing Values by Column:")
    for column, count in df.isnull().sum().items():
        report_lines.append(f"{column}: {count}")

    numeric_df = df.select_dtypes(include="number")

    if not numeric_df.empty:
        report_lines.append("\nNumerical Column Averages:")
        for column, average in numeric_df.mean().items():
            report_lines.append(f"{column}: {average:.2f}")

    report_lines.append("\nAnalysis Summary:")
    report_lines.append(
        "Descriptive statistics and visualizations were generated "
        "to explore the student dataset."
    )
    report_lines.append(
        "These results describe the available data and do not "
        "independently validate prediction accuracy."
    )

    report_text = "\n".join(report_lines)

    output_path = Path.cwd() / "day27_student_performance_report.txt"
    output_path.write_text(report_text, encoding="utf-8")

    print(report_text)
    print("\nReport saved successfully!")
    print("File path:", output_path.resolve())

else:
    print("Dataset is not loaded. Please run Cell 2 first.")
```
